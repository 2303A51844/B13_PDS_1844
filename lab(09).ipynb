```python
# Import necessary libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd # Import pandas

# Load the Iris dataset
# Use pandas to read the CSV file directly
iris = pd.read_csv("iris_dataset.csv")

# Display first few rows
print(iris.head())

# Create a pair plot
sns.pairplot(
    iris,
    hue="species",         # Color code by species
    diag_kind="hist",      # Show histograms on the diagonal
    palette="husl"         # Use a colorful palette
)

# Show the plot
plt.show()
```

```python
# Import necessary libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd



# Option 2 (if you have a CSV file):
cars = pd.read_csv("car_dataset_sample.csv")

# Inspect first few rows
print(cars.head())

# Drop missing values for clean plotting
cars = cars.dropna(subset=["horsepower", "weight", "mpg"])

# Create a pair plot
sns.pairplot(
    cars,
    vars=["horsepower", "weight", "mpg"],  # Select relevant numerical features
    hue="origin",                           # Color by car origin (usa, europe, japan)
    diag_kind="hist",                       # Histograms on the diagonal
    palette="coolwarm"
)

# Show the plot
plt.show()
```
